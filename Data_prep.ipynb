```python
from datetime import datetime
# Function to convert French month names to datetime format
def convert_to_datetime(french_date):
    if isinstance(french_date, str):  # Only proceed if the value is a string
        month_mapping = {
            'janvier': 'January', 'février': 'February', 'mars': 'March', 'avril': 'April', 'mai': 'May', 'juin': 'June',
            'juillet': 'July', 'août': 'August', 'septembre': 'September', 'octobre': 'October', 'novembre': 'November',
            'décembre': 'December'
        }
        for french_month, english_month in month_mapping.items():
            if french_month in french_date:
                french_date = french_date.replace(french_month, english_month)
                break
        return datetime.strptime(french_date, '%B %Y')
    return french_date  # Return the original value if it's not a string

# Apply the function to convert the entire column to datetime
df_france_fr['stay date'] = df_france_fr['stay date'].apply(convert_to_datetime)
```

```python
import re

# Function to extract the number of nights from the text
def extract_nights(text):
    if isinstance(text, str):
        # Extract the first number found in the text
        match = re.search(r'\d+', text)
        if match:
            return int(match.group(0))  # Return the number as an integer
    return None  # Return None if no number is found

# Apply the function to the column
df_france_fr['Number of Nights'] = df_france_fr['Number of Nights'].apply(extract_nights)
```

```python
import spacy
import re
from nltk.corpus import stopwords
from spellchecker import SpellChecker

# Charger le modèle large de spaCy
nlp = spacy.load('fr_core_news_lg')

# Stopwords enrichis
stop_words = set(stopwords.words('french')).union(nlp.Defaults.stop_words)


# Expansion des formes contractées
def expand_contractions(text):
    contractions = {
        "l'": "le ",
        "d'": "de ",
        "j'": "je ",
        "qu'": "que ",
        "c'": "ce ",
    }
    for contraction, expanded in contractions.items():
        text = text.replace(contraction, expanded)
    return text


# Prétraitement avec lemmatisation améliorée
def preprocess_text(text):
    if isinstance(text, str):
        text = text.lower()
        text = re.sub(r'\W+|\d+', ' ', text)
        doc = nlp(text)
        tokens = [token.lemma_ for token in doc if token.lemma_ not in stop_words and token.pos_ in ['NOUN', 'VERB', 'ADJ']]
        return tokens
    return []

```

```python
# Function to count occurrences of keywords in each row for specified columns.
def count_keywords_per_row(dataframe, columns, keywords_dict):
    # Prepare a new dataframe to store counts for each topic and row
    results = pd.DataFrame(index=dataframe.index)
    for topic, keywords in keywords_dict.items():
        for col in columns:
            col_name = f"{topic}_{col}"
            results[col_name] = dataframe[col].apply(lambda x: sum(1 for word in x if word in keywords))
    return results

# Apply the function to count keywords for each row.
row_keyword_counts = count_keywords_per_row(df_france_fr, columns_to_search, topics_keywords)

# Merge the results back into the original dataframe for reference.
df_with_keyword_counts = pd.concat([df_france_fr, row_keyword_counts], axis=1)
```

```python
import pandas as pd

# Trier les données par hôtel et date
df_sorted = df_econometrie.sort_values(by=['Hotel Link', 'review_date_clean'])

# Liste des tailles d'échantillons à extraire
sample_sizes = [10, 20, 30, 40, 50]

# Liste des colonnes numériques à sommer (exclure celles qui doivent être moyennées)
sum_columns = df_sorted.select_dtypes(include=['number']).columns.tolist()
sum_columns.remove("rating_value")


# Création d'un DataFrame pour stocker les résultats
df_aggregated = []

# Appliquer l'agrégation pour chaque hôtel et chaque échantillon de commentaires
for size in sample_sizes:
    df_grouped = (
        df_sorted.groupby('Hotel Link').head(size)  # Prendre les X premiers commentaires par hôtel
        .groupby('Hotel Link')  # Regrouper à nouveau par hôtel
        .agg({**{col: 'sum' for col in sum_columns},  # Somme des colonnes sélectionnées
              "rating_value": "mean",  # Moyenne du rating
              })  # Moyenne du note
        .reset_index()
    )
    
    df_grouped['Sample_Size'] = size  # Ajouter une colonne indiquant la taille de l'échantillon
    df_aggregated.append(df_grouped)  # Ajouter au résultat

# Concaténer tous les résultats
df_final_1 = pd.concat(df_aggregated, ignore_index=True)


```
